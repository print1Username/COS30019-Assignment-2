## Matplotlib

Before we create the AI model, we can create the function for the Chart generator.
\
First, we can try to import the requirement package.

````python
import os
import matplotlib.pyplot as plt

os.environ['KMP_DUPLICATE_LIB_OK'] = 'TRUE'
````

After importing the required packages, we can create a class called **Matplolib** and set the class title, size, and other details.
\
\
For the width and height, it means the width and height for the diagram. For the color of the chart, I setting the default is blue.

````python
class Matplotlib:
    def __init__(self, width, height, title, x_label, y_label, color = "blue", grid = True, alpha = 1):
        self.title = title
        self.width = width # The width for the diagram
        self.height = height # The height for the diagram
        self.x_label = x_label
        self.y_label = y_label
        self.color = color # The color for the chart (Default Blue)
        self.grid = grid # Generate the grid in the diagram (Default: True)
        self.alpha = alpha # The shade for the grid line (Default: 1)

        self.dictionary = {}
        self.x_list = []
        self.y_list = []
````

After creating the classes, we can have a function for the data list.
\
In this program, there have two type of the data list, that is a list and a dictionary.
\
For the function, that have two type for adding data, one is just adding the value inside the data, and other one is combine two list into one.

````python
    def AddData(self, x_value, y_value):
        self.x_list.append(x_value)
        self.y_list.append(y_value)

    def AddDataList(self, x_value_list, y_value_list):
        self.x_list += x_value_list
        self.y_list += y_value_list

    def ClearData(self):
        self.x_list = []
        self.y_list = []
        self.dictionary = {}

    def AddDictionary(self, x_value, y_value):
        self.dictionary[x_value] = y_value

    def AddDictionaryList(self, dictionary):
        self.dictionary = dictionary.copy()
````

For the dictionary, its cannot be read by matplotlib, we need to conver the dictionary to list to reading by matplolib.

```
Convert the dictionary to a list

dictionary = {

}

After Convert:
```

````python
    def ConvertDictionaryToList(self):
        self.x_list = []
        self.y_list = []

        for dict in self.dictionary:
            self.x_list.append(dict)
            self.y_list.append(self.dictionary[dict])
````

Next, we can make a function for create the chart and save it into image.
\
This function can generate four type of the chart, you can setting the chart that you want to generate.
- 1: Line Chart
- 2: Scatter Chart
- 3: Bar Chart
- 4: Pie Chart

For the funtion, it will generate the windows by default, if you want to disable the function, you need to set the **window = False**.
\
For the saving, it will default save the image with "jpg" format. You can change it to other format just like "jpg", jpeg"", "png", "pdf".
\
The function have some of the safety, if the diagram have some error when generate or saving, it will print out the error details.

````python
    def CreateChart(self, diagram_type, window = True, save_format = "jpg", save_directory = None):
        try:
            plt.figure(figsize=(self.width, self.height))
            plt.title(self.title)
            plt.xlabel(self.x_label)
            plt.ylabel(self.y_label)
            plt.grid(self.grid, alpha=self.alpha)

            if window is not True:
                plt.ioff()

            match diagram_type:
                case 1: # Line Chart 线状图
                    plt.plot(self.x_list, self.y_list, color = self.color)
                case 2: # Scatter Chart 散点图
                    plt.scatter(self.x_list, self.y_list, color = self.color, s = 50)
                case 3: # Bar Chart 柱状图
                    plt.bar(self.x_list, self.y_list, color = self.color)
                case 4: # Pie Chart 饼图
                    colors = ['gold', 'yellowgreen', 'lightcoral', 'lightskyblue']
                    plt.pie(self.y_list, labels=self.x_list, colors=colors, autopct='%1.1f%%', startangle=90)
                    plt.axis('equal')
                case _:
                    print("ERROR: Invalid chart type.")

            save_format_list = ["jpg", "jpeg", "png", "pdf"]

            if save_format.lower() not in save_format_list:
                print(f"⚠️ ERROR: Can't save the file with '{save_format}'. Supported formats: {save_format_list}")
                return False

            try:
                # If the save directory have value, it will save in the directory
                # else it will save as the directory same as the file
                if save_directory:
                    os.makedirs(save_directory, exist_ok=True)
                    file_path = os.path.join(save_directory, f"{self.title}.{save_format.lower()}")
                else:
                    file_path = f"{self.title}.{save_format.lower()}"

                plt.savefig(file_path, bbox_inches='tight', dpi=300)
                plt.close()

                print(f"Chart saved successfully: {file_path}")
                return True

            # Output the error if the chart cannot save
            except Exception as e:
                print(f"⚠️ ERROR: Failed to save chart: {str(e)}")
                return False

        except Exception as e:
            print(f"⚠️ ERROR for the chart: {str(e)}")
````

## Swin Transformer

For the Swin Transformed, we can import the package that we need.

````python
import os
import torch

import torch.nn as nn
import torchvision.models as models
import torchvision.transforms as transforms

from PIL import Image
from torch.utils.data import DataLoader
from torchvision.datasets import ImageFolder
````

After impor the requirement, we can start to create a class name SwinTransformed.
\
For the device, it will check the device is using CPU or GPU.

````python
class SwinTransformer:
    def __init__(self, image_size, model_path):
        # Check the device for the AI training
        self.device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

        self.model_path = model_path

        self.transform = transforms.Compose([
            transforms.Resize((image_size, image_size)),
            transforms.ToTensor(),
            transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
        ])
````

After creating the classes, we can prepared the model using.

````python
    def PrepareModel(self, num_classes):
        print("Preparing SwinTransformer Model")

        self.model = models.swin_b(weights='IMAGENET1K_V1')
        self.model.head = nn.Linear(self.model.head.in_features, num_classes)

        # Put the model to the device
        self.model.to(self.device)
        print("Model prepared successfully.")
````

After prepared model, we need to load the data for training and validation.
\
⚠️ Reminder again: Change the file location at above.

````python
    def LoadData(self, data_path, split, batch_size):
        print(f"Load data from {data_path}")

        # Create the dataset
        dataset = ImageFolder(data_path, transform=self.transform)

        # Save the classes name
        self.class_names = dataset.classes
        print(f"Finding classes: {self.class_names}")
        print(f"Total have {len(dataset)} images.")

        # Split dataset to training and testing
        train_size = int(split * len(dataset))
        test_size = len(dataset) - train_size
        train_data, test_data = torch.utils.data.random_split(dataset, [train_size, test_size])

        # Creating a Data Loader - Batch Processing Images
        self.train_loader = DataLoader(train_data, batch_size=batch_size, shuffle=True)
        self.test_loader = DataLoader(test_data, batch_size=batch_size, shuffle=False)

        print(f"Training data: {train_size} images.")
        print(f"Testing data: {test_size} images.")

        return len(self.class_names)
````

After load the dataset, we can start to train and validation our model

````python
    def TrainModel(self, epochs):
        print("Start training model...")

        # Setting the optimizer and loss function
        optimizer = torch.optim.Adam(self.model.parameters(), lr=0.001)
        criterion = nn.CrossEntropyLoss()

        chart = Matplotlib(8,6,'Training Accuracy','Epoch','Accuracy')

        # Start training loop
        for epoch in range(epochs):
            print(f"\nTraining {epoch + 1}/{epochs} round...")

            # Train mode
            self.model.train()
            train_correct = 0
            train_total = 0

            # Process each batch of training data
            for batch_idx, (images, labels) in enumerate(self.train_loader):
                images, labels = images.to(self.device), labels.to(self.device)  # Putting data on GPU/CPU
                optimizer.zero_grad()  # Clear gradient
                outputs = self.model(images)  # Forward propagation
                loss = criterion(outputs, labels)  # Calculate loss
                loss.backward()
                optimizer.step()  # Backpropagation

                _, predicted = outputs.max(1)
                train_total += labels.size(0)
                train_correct += predicted.eq(labels).sum().item()  # Calculate accuracy

                # Display progress every batches
                print(f">\tBatch {batch_idx + 1}/{len(self.train_loader)}, Loss: {loss.item():.4f}")

            train_accuracy = 100. * train_correct / train_total  # Calculate training accuracy
            print(f"Training Accuracy:\t{train_accuracy:.2f}%")
            self.TestModel()

            # Add the epoch and accuracy to the chart
            chart.AddData(epoch,train_accuracy)

        chart.CreateChart(3,False,"png","results")
        self.SaveModel()
````

Next, it will test the model and saving the model.
\
For saving the model, it will save the model name that you set before.

````python
    def TestModel(self):
        self.model.eval()
        test_correct = 0
        test_total = 0

        with torch.no_grad():  # No gradient calculation is required during testing
            for images, labels in self.test_loader:
                images, labels = images.to(self.device), labels.to(self.device)
                outputs = self.model(images)
                _, predicted = outputs.max(1)
                test_total += labels.size(0)
                test_correct += predicted.eq(labels).sum().item()

        test_accuracy = 100. * test_correct / test_total
        print(f"Testing accuracy: {test_accuracy:.2f}%")
````

````python
    def SaveModel(self):
        torch.save({
            'model': self.model.state_dict(),
            'classes': self.class_names
        }, 'animal_model.pth')
````

After we setting the model, we can start for AI model Detection.
\
Before AI model detection, we need to load the model before we save it.

````python
    def LoadModel(self):
        print(f"Load Model: {self.model_path}")
        checkpoint = torch.load(self.model_path, map_location=self.device)

        # Rebuilding the model structure
        num_classes = len(checkpoint['classes'])
        self.PrepareModel(num_classes)

        # Load model weights and class names
        self.model.load_state_dict(checkpoint['model'])
        self.class_names = checkpoint['classes']
        self.model.eval()
        print("Model loading completed")
````

After we have load the model that we save, we can start to Predict the image:
\
It will return the classes and the confidence.

````python
    def PredictImage(self, image_path):
        # Load Image
        image = Image.open(image_path).convert('RGB')
        image_tensor = self.transform(image).unsqueeze(0).to(self.device)

        self.model.eval()
        with torch.no_grad():
            outputs = self.model(image_tensor)
            probabilities = torch.softmax(outputs[0], dim=0)
            predicted_class_idx = torch.argmax(probabilities).item()
            confidence = probabilities[predicted_class_idx].item()

        predicted_animal = self.class_names[predicted_class_idx]
        return predicted_animal, confidence
````

If we want to Predict with a folder, we can create a function for loop the function above.

````python
    def PredictFolder(self, folder_path, result_folder = ""):
        print(f"Analyzing folder: {folder_path}")

        # Supported image format
        image_formats = ['.jpg', '.jpeg', '.png']

        results = []
        chart = Matplotlib(8,8,'Detecting Animal Chart','Animal','')
        dict = {}
        for filename in os.listdir(folder_path):
            # Check if it is an image file
            if any(filename.lower().endswith(fmt) for fmt in image_formats):
                image_path = os.path.join(folder_path, filename)
                try:
                    animal, confidence = self.PredictImage(image_path)
                    results.append({
                        'filename': filename,
                        'predicted_animal': animal,
                        'confidence': confidence
                    })

                    if animal in dict:
                        num = dict[animal]
                        dict.pop(animal)
                        dict[animal] = num + 1
                    else:
                        dict[animal] = 1

                    print(f"📷 {filename} -> {animal} (Accuracy: {confidence:.2%})")
                except Exception as e:
                    print(f"❌ {filename} Handling failure: {e}")

        chart.AddDictionaryList(dict)
        chart.ConvertDictionaryToList()
        chart.CreateChart(4,save_format="png",save_directory=result_folder)
        return results
````
